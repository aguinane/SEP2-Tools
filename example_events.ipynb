## Generate some example control events and save to DB

```python
from sep2tools.event_examples import example_controls, example_default_control
from sep2tools.events_db import add_events, cleanup_events

program = "EXAMPLEPRG"
example_events = [
    *example_controls(program=program),
    example_default_control(program=program),
]
add_events(example_events)
cleanup_events()
```

## Load those events back into a dataframe

Any superseded events will be automatically handled. 
Non-overlapping superseded are handled by the cleanup function.
Any overlapping events are handled when reading back in and split up as required. 

```python
import polars as pl

from sep2tools.events_clean import get_mode_event_values
from sep2tools.events_db import get_program_modes

modes = get_program_modes(program=program)
mode = modes[0]
print(f"Events for mode {mode}:")
event_values = get_mode_event_values(program=program, mode=mode, retro_hours=12)
df = pl.DataFrame(event_values)
print(df)
```

## Chart the data

```python
import altair as alt

chart = (
    alt.Chart(df)
    .mark_line()
    .encode(x="ts:T", y="value:Q")
    .properties(title=f"Events for {program} - {mode}")
)
chart.show()
```
